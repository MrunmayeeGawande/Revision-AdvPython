### Multiplying in a list Vs an array

```python
# If you direcltly multiply by 4 to a list , list will repeat itself
a = [5, 6, 7, 8, 9, 10]
print(4 * a)
```

```python
1.5*a
```

```python
a = [5, 6, 7, 8, 9, 10]
print(a)

# Intialize blank results list
r = []

# Apply for loop on a
for i in a:
    print(f"{i} x 4 = {4 * i}")
    r.append(4 * i)
```

```python
print(r)
```

```python
a1 = np.array(a)
print(a1, type(a1))
b1 = 4 * a1
print(b1)
```

## Sum of elements inside a list vs array

```python
# List
b = [4, 5, 6, 7, 8, 9, 10, 11, 12]
print(b)

# Intialuze sum with 0 
s = 0

for i in b:
    s = s + i

print(f"Sum of elements inside list {b} is {s}")
```

```python
# Array
b1 = np.array(b)
b1.sum()
```

```python
b1.prod()
```

```python
b1.mean()
```

```python
b1.max()
```

```python
b1.min()
```

### Simulate dice throw betwee 1 to 6

```python
height = np.random.normal(loc=5.5, scale=0.5/3, size=100000)
print(height)
print(height.shape)
```

### Whenever providing angles it should be in raidians
1 radian = pi / 180 degree

```python
ang_deg = [0, 30, 45, 60, 90, 120, 180, 270, 360]
ang_rad = np.radians(ang_deg)
print(ang_rad)
```

```python
np.sin(ang_rad)
```

```python
np.cos(ang_rad)
```

```python
np.tan(ang_rad)
```
